```python
from pathlib import Path
from collections import Counter
import re

FILES = [
    "data/PKU_TianWang/repo_file_list.cache.txt",
    "data/PKU_TianWang2/repo_file_list.cache.txt",
    "data/PKU_TianWang3/repo_file_list.cache.txt",
]

PATTERN = re.compile(r'^[A-Z](\d{4})\d{2}/Web_Raw\.[A-Z]\d{6}\.\d+$')


def count_years(files):
    counter = Counter()
    total = 0

    for file_path in files:
        path = Path(file_path)
        with path.open("r", encoding="utf-8", errors="ignore") as f:
            for line in f:
                line = line.strip()
                if not line:
                    continue

                m = PATTERN.match(line)
                if m:
                    year = m.group(1)
                    counter[year] += 1
                    total += 1

    return total, counter


if __name__ == "__main__":
    total, year_counts = count_years(FILES)

    print(f"总数: {total}")
    print("各年份数量：")
    for year in sorted(year_counts):
        print(f"{year}: {year_counts[year]}")
```

```python
from pathlib import Path
from collections import Counter
import re

def count_web_raw_files_by_year(root_dir):
    root = Path(root_dir)
    year_counter = Counter()

    for file in root.rglob("Web_Raw*"):
        if file.is_file() and not file.name.endswith((".metadata", ".lock")):
            parent_name = file.parent.name
            match = re.search(r"[A-Z](\d{4})\d{2}", parent_name)
            if match:
                year = int(match.group(1))
                year_counter[year] += 1
            else:
                year_counter["UNKNOWN"] += 1

    return year_counter

year_counts = count_web_raw_files_by_year("data")
total = sum(v for v in year_counts.values() if isinstance(v, int))

print(f"总数量: {total}")
for year in sorted(k for k in year_counts if isinstance(k, int)):
    print(f"{year}: {year_counts[year]}")

if "UNKNOWN" in year_counts:
    print(f"UNKNOWN: {year_counts['UNKNOWN']}")
```

```python
from pathlib import Path

BASE_DIRS = [
    "outputs/PKU_TianWang/raw_data",
    "outputs/PKU_TianWang2/raw_data",
    "outputs/PKU_TianWang3/raw_data",
]

def count_html_files(base_dirs):
    total = 0
    per_dir = {}

    for base_dir in base_dirs:
        path = Path(base_dir)
        count = sum(1 for p in path.rglob("*") if p.suffix.lower() == ".html")
        per_dir[base_dir] = count
        total += count

    return total, per_dir


if __name__ == "__main__":
    total, detail = count_html_files(BASE_DIRS)

    print("Per-directory counts:")
    for k, v in detail.items():
        print(f"{k}: {v}")

    print(f"\nTotal .html files: {total}")
```

```python
from pathlib import Path

def count_web_raw_files(root_dir):
    root = Path(root_dir)
    count = 0

    for file in root.rglob("Web_Raw*"):
        if file.is_file() and not file.name.endswith((".metadata", ".lock")):
            count += 1
    return count

print(count_web_raw_files("data"))
```
